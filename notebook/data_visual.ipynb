```python
lack_depth:list = []
lack_dim:list = []
for x in val_image_path_list:
    img = plt.imread(x)
    if(len(img.shape) != 3):
        # print(f"This image does not have depth {x}\n")
        lack_depth.append(x)
    if(img.shape[0]<448 or img.shape[1]<448):
        # print(f"This image does not have enough dimensions {x}\n")
        lack_dim.append(x)
print("Done !")
```
